Cleaning Data



1.   Karena missing cukup banyak dan kolom bersifat kategorikal maka diisi dengan 'Tidak_Diketahui' supaya data tidak hilang dan model tetap bisa belajar pola
2.   Drop kolom ID karena tidak berguna untuk model


1.   Hapus baris yang mengandung outlier karena Outlier hanya 4 data dan Dataset cukup besar (~1000)





```python
# Kolom yang memiliki missing value
kolom_missing = ['Jenis_Penyakit', 'Pantangan_Makanan', 'Alergi']

# Mengisi dengan label khusus
for kolom in kolom_missing:
    df[kolom] = df[kolom].fillna('Tidak_Diketahui')
print(df.isnull().sum())
```

```python
#hapus kolom ID
df = df.drop(columns=['ID_Pasien'])
```

```python
# Hitung IQR untuk kolom BMI
Q1 = df['Indeks_Massa_Tubuh'].quantile(0.25)
Q3 = df['Indeks_Massa_Tubuh'].quantile(0.75)
IQR = Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

# Filter data tanpa outlier
df_clean = df[
    (df['Indeks_Massa_Tubuh'] >= batas_bawah) &
    (df['Indeks_Massa_Tubuh'] <= batas_atas)
]

print("Jumlah data sebelum:", df.shape[0])
print("Jumlah data setelah hapus outlier:", df_clean.shape[0])
```

Feature Engineering

```python
#Kategori BMI (lebih informatif dari angka mentah)
def kategori_bmi(bmi):
    if bmi < 18.5:
        return 'underweight'
    elif bmi < 25:
        return 'normal'
    elif bmi < 30:
        return 'overweight'
    else:
        return 'obese'

df['Kategori_BMI'] = df['Indeks_Massa_Tubuh'].apply(kategori_bmi)
```

```python
#Kategori Usia
def kategori_usia_custom(usia):
    if usia <= 29:
        return '19-29 tahun'
    elif usia <= 49:
        return '30-49 tahun'
    elif usia <= 64:
        return '50-64 tahun'
    elif usia <= 80:
        return '65-80 tahun'
    else:
        return '80+ tahun'

df['Kategori_Usia'] = df['Usia'].apply(kategori_usia_custom)
```

```python
#Status Aktivitas Fisik (numerik → kategorikal sederhana)
def aktivitas_level(jam):
    if jam < 3:
        return 'rendah'
    elif jam < 7:
        return 'sedang'
    else:
        return 'tinggi'

df['Kategori_Olahraga'] = df['Jam_Olahraga_per_Minggu'].apply(aktivitas_level)
```

```python
#Risiko Kesehatan
def risiko(row):
    skor = 0

    if row['Glukosa_mg/dL'] > 140:
        skor += 1
    if row['Kolesterol_mg/dL'] > 200:
        skor += 1
    if row['Tekanan_Darah_mmHg'] > 140:
        skor += 1

    if skor >= 2:
        return 'tinggi'
    elif skor == 1:
        return 'sedang'
    else:
        return 'rendah'

df['Risiko_Kesehatan'] = df.apply(risiko, axis=1)
```

Encoding Kolom Kategorikal

```python
df_encoded = df.copy()
#Label Encoding (Binary)

df_encoded['Jenis_Kelamin'] = df_encoded['Jenis_Kelamin'].map({
    'Male': 1,
    'Female': 0
})

# One Hot Encoding
kolom_ohe = [
    'Jenis_Penyakit',
    'Tingkat_Keparahan',
    'Tingkat_Aktivitas_Fisik',
    'Pantangan_Makanan',
    'Alergi',
    'Preferensi_Masakan',
    'Kategori_BMI',
    'Kategori_Usia',
    'Kategori_Olahraga',
    'Risiko_Kesehatan'
]

df_encoded = pd.get_dummies(df_encoded, columns=kolom_ohe, drop_first=True)

# Encode Target
df_encoded['Rekomendasi_Diet'] = df_encoded['Rekomendasi_Diet'].map({
    'Balanced': 0,
    'Low_Carb': 1,
    'Low_Sodium': 2
})

# Cek hasil
print(df_encoded.head())
print("\nShape:", df_encoded.shape)
```

```python
# Ambil hanya kolom boolean
bool_cols = df_encoded.select_dtypes(include='bool').columns

# Ubah hanya kolom boolean ke int
df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)
df_encoded.head()
```

Pisahkan Kolom Sebelum Scaling

```python
X = df_encoded.drop(columns=['Rekomendasi_Diet'])
y = df_encoded['Rekomendasi_Diet']
```

Scaling Kolom Numerik

```python
# Pisahkan kolom numerik & kategori
num_cols = X.select_dtypes(include=['int64', 'float64']).columns

# KECUALIKAN kolom dummy (yang berasal dari OHE)
# biasanya kolom dummy itu banyak dan berupa 0/1

# Cara aman: pilih hanya kolom numerik asli
num_cols = [
    'Usia',
    'Berat_Badan_kg',
    'Tinggi_Badan_cm',
    'Indeks_Massa_Tubuh',
    'Asupan_Kalori_Harian',
    'Kolesterol_mg/dL',
    'Tekanan_Darah_mmHg',
    'Glukosa_mg/dL',
    'Jam_Olahraga_per_Minggu',
    'Kepatuhan_Terhadap_Diet',
    'Skor_Ketidakseimbangan_Nutrisi'
]

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = X.copy()
X_scaled[num_cols] = scaler.fit_transform(X[num_cols])
```

Output akhir

```python
# Simpan fitur
X_scaled.to_csv('diet_recommendations_cleaned_scaled.csv', index=False)

# Simpan label saja
y.to_csv('diet_recommendations_labels.csv', index=False)

print("File berhasil disimpan!")
```
